```python
import pandas as pd
import numpy as np
```

```python
# Đọc dữ liệu
df = pd.read_csv('project_heart_disease.csv')
display(df.info())
display(df.head())

missing_values = df.isnull().sum()
total_missing_values = missing_values.sum()

# Đếm số lượng giá trị thiếu trong các cột số
numeric_missing = df.select_dtypes(include=[np.number]).isnull().sum()

# Đếm số lượng giá trị thiếu trong các cột chuỗi
object_missing = df.select_dtypes(include=['object']).isnull().sum()

# In kết quả
print("Missing values per column:\n", missing_values)
print(f"Total missing values: {total_missing_values}")

# Xử lý giá trị thiếu
for col in df.select_dtypes(include=[np.number]).columns:
    df[col].fillna(df[col].mean(), inplace=True)

for col in df.select_dtypes(include=['object']).columns:
    df[col].fillna('None', inplace=True)

#Xử lí giá trị trùng lặp
duplicate_count = df.duplicated().sum()
print(f"Total duplicate values:{duplicate_count}")
df = df.drop_duplicates()

#Xuất ra file mới
df.to_csv('Preprocessing_project_heart_disease.csv', index = False)

# Kiểm tra lại giá trị thiếu sau khi xử lý
missing_values_after = df.isnull().sum()
missing_values_count_after = missing_values_after.sum()
print(f"Total missing values after processing: {missing_values_count_after}")

# Kiểm tra lại giá trị trùng lặp sau khi xử lý
duplicate_count_after = df.duplicated().sum()
print(f"Total duplicate values after processing: {duplicate_count_after}")
```

```python
# Lưu bản sao trước khi xử lý giá trị thiếu
# Đọc dữ liệu từ file CSV
df1 = pd.read_csv('project_heart_disease.csv')

# Chuyển các giá trị rỗng hoặc "NULL" thành NaN để xử lý nhất quán
df1.replace(["", "NULL"], pd.NA, inplace=True)

# Tạo bản sao của dữ liệu gốc
df_original = df1.copy()

# Tổng số dòng trong dataset
total_rows = len(df_original)

# Tạo danh sách chứa dữ liệu profiling
profile_data = []

# Duyệt qua từng cột trong dataset
for column in df_original.columns:
    # Đếm số lượng giá trị NULL
    null_count = df_original[column].isna().sum()

    # Missing count (đặt bằng 0 vì pandas đã xử lý khi đọc)
    missing_count = 0

    # Số lượng giá trị thực tế (không phải NULL)
    actual_count = total_rows - null_count

    # Tính độ đầy đủ của dữ liệu (%)
    completeness = (actual_count / total_rows) * 100 if total_rows > 0 else 0

    # Tính cardinality: số lượng giá trị duy nhất (bỏ qua NaN)
    cardinality = len(df_original[column].dropna().unique())

    # Tính độ duy nhất (%): tỷ lệ giá trị duy nhất so với tổng số dòng
    uniqueness = (cardinality / total_rows) * 100 if total_rows > 0 else 0

    # Tính độ khác biệt (%): tỷ lệ giá trị duy nhất so với số giá trị thực tế
    distinctness = min((cardinality / actual_count) * 100, 100) if actual_count > 0 else 0

    # Lưu thông tin vào danh sách
    profile_data.append([column, null_count, missing_count, actual_count, completeness, cardinality, uniqueness, distinctness])

# Chuyển danh sách thành DataFrame
summary = pd.DataFrame(profile_data, columns=["Field Name", "NULL", "Missing", "Actual", "Completeness (%)", "Cardinality", "Uniqueness (%)", "Distinctness (%)"])

# Hiển thị bảng kết quả
display(summary)
```
